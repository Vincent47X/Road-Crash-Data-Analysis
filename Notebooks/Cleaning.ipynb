```python
#installing required libraries:
import numpy as np
import pandas as pd
import matplotlib as plot
```

```python
#Reading data:
df = pd.read_csv('Raw_data/Crash_Report.csv')
```

```python
#Raw data:
df.shape

df.info()

df.head()
```

```python
#Checking for NULL values:
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_pct
}).head(15)
```

```python
#Handling NULL values:
cat_cols = df.select_dtypes(include="object").columns

df[cat_cols] = df[cat_cols].fillna("Unknown")
```

```python
#Validating NULL values:
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_pct
}).head(15)
```

```python
#Replacing the 0 value with Nan:
df["Vehicle Year"] = df["Vehicle Year"].replace(0, np.nan)
df = df.dropna(subset=["Vehicle Year"])
```

```python
#Fixing date and time column:
df["Crash Date/Time"] = pd.to_datetime(
    df["Crash Date/Time"],
    errors="coerce"
)

df = df.dropna(subset=["Crash Date/Time"])
```

```python
df.head()
```

```python
#Checking for duplicate values:
df.duplicated().sum()
```

```python
#Droping duplicate values:
df = df.drop_duplicates()
```

```python
df["Crash Hour"] = df["Crash Date/Time"].dt.hour
df["Crash Day"] = df["Crash Date/Time"].dt.day_name()
df["Crash Month"] = df["Crash Date/Time"].dt.month
```

```python
#Converting texts/objects to proper format:
columns_to_clean = [
    "Injury Severity",
    "Vehicle Body Type",
    "Vehicle Make",
    "Crash Day"
]

for col in columns_to_clean:
    df[col] = (
        df[col].str.strip()
        .str.lower()
        .str.title()
    )
```

```python
df.info()
```

```python
#Validating:
df.head()
```

```python
#Savind the cleaned report in a csv file:
df.to_csv("Crash_Report_Cleaned.csv", index=False)
```
